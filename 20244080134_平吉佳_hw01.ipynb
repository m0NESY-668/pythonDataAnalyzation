## 欧几里得算法

```python

def gcd(m,n):
  if n == 0:
    return m
  else:
    return gcd(n, m % n)
print(gcd(5040,60))
```

## 近似欧拉数e

```python
def euler_limit(n):
  e = (1+1/n)**n
  return e
```

```python
def euler_infinite_sum(n):
  if n == 0:
    return 0
  sum_e = 0
  a = 1
  for i in range(n):
    sum_e += 1/a
    a *= (i+1)
  return sum_e
```

```python
import math
def euler_approx(ɛ):
  ture_e = math.exp(1)
  sum_e = 0
  fact = 1
  k = 0
  while True:
    term = 1/fact
    sum_e += term
    if abs(sum_e - ture_e) < ɛ:
      break
    k += 1
    fact *= k
  return sum_e
```

```python
def print_euler_sum_table(n):
    for i in range(1, n + 1):
        print(euler_infinite_sum(i))


print(euler_infinite_sum(0)) # 0.0
print(euler_infinite_sum(1)) #1.0
print(euler_infinite_sum(4)) #≈2.6666666
print(euler_limit(1000))
print(euler_approx(0.001))
print_euler_sum_table(5)
```

## 测试整数属性

```python
def is_power_of_2(n):
  if n <= 0:
    return False
  while True:
    a = n%2
    if a == 1:
      return n==1
    else:
      n = n/2

print(is_power_of_2(1))   # True (2^0)
print(is_power_of_2(8))   # True
print(is_power_of_2(6))   # False
print(is_power_of_2(16))  # True
print(is_power_of_2(0))   # False
print(is_power_of_2(-4))  # False
print(is_power_of_2(7))   # False
```

```python
def is_power(b,n):
    if b <=0:
      return False
    while True:
       a = n%b
       if a == 1:
          return n==1
       else:
          n = n/b


print(is_power(2,1))   # True (2^0)
print(is_power(2,8))   # True
print(is_power(2,10))   # False
print(is_power(3,6))   # False
print(is_power(3,9))  # True
print(is_power(3,27))   # True
```
