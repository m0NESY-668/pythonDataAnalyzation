```python

def score_level(score):

    # 检查成绩是否在有效范围内
    if score >= 90 and score <= 100:
        return "优秀"
    elif score >= 80 and score < 90:
        return "良好"
    elif score >= 60 and score < 80:
        return "及格"
    elif score >= 0 and score < 60:
        return "不及格"
    else:
        raise ValueError("ValueError: score must be between 0 and 100")
print(score_level(90))
print(score_level(80))
print(score_level(60))
print(score_level(50))
#print(score_level(200))
```

```python
# 编写一个递归版的阶乘函数
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))  
```

```python
# 编写一个迭代版的阶乘函数
def factorial2(n):
  if n ==0 or n == 1:
    return 1
  else:
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial2(0))  
```

```python
def summary(values):
  if len(values) == 0:
    raise ValueError("列表不能是空的")
  
  return max(values), min(values), sum(values)/len(values)


result = summary([85, 92, 78, 90, 88])
print(result) 

minimum, maximum, average = summary([85, 92, 78, 90, 88])
print(f"最小值: {minimum}")
print(f"最大值: {maximum}")
print(f"平均值: {average:.2f}")
```

```python
import os

cwd = os.getcwd()
print(f"当前工作目录是{cwd}")


data_dir = os.path.join("..","exp01_data")
file_path = os.path.join(data_dir,"students.txt")
print(f"[数据目录]     {os.path.abspath(data_dir)}")
print(f"[文件路径]     {os.path.abspath(file_path)}")

os.makedirs(data_dir,exist_ok=True)
print("[目录创建] 成功")

```

```python
data_lines = [
    "name,score",
    "Alice,85",
     "Bob,58",
    "Carol,92",
    "David,76",
    "Eva,60",
    "Frank,89",
    "Grace,abc",
    "Henry,105"
]

with open(file_path,"w",encoding="utf-8") as f:
   for line in data_lines:
        f.write(line + "\n")
print("[文件写入] 成功")    
with open(file_path,"r",encoding="utf-8") as f:
    lines = f.readlines()
    
    
for line in lines[1:]:  # 从第2行开始（跳过表头）
    line = line.strip()  # 去除首尾空白和换行符
    if not line:         # 跳过空行
        continue
    name, score = line.split(",")  # 用逗号分割字符串
    print(f"姓名: {name:<8} 成绩: {score}")
```

```python
def parse_student(line):
    """
    解析学生记录字符串，将成绩转换为整数，并验证数据有效性。
    
    参数:
        line: 包含姓名和成绩的字符串，用逗号分隔，例如 "Alice,85"
    
    返回:
        tuple: (name, score) 姓名字符串和成绩整数
    
    抛出:
        ValueError: 当字段数量不对、姓名为空、或成绩不在 0~100 范围内时
    """
    # 去除首尾空白
    line = line.strip()
    
    # 按逗号分割字段
    parts = line.split(",")
    
    # 检查字段数量是否为 2
    if len(parts) != 2:
        raise ValueError(f"字段数量错误: 期望 2 个字段，实际得到 {len(parts)} 个")
    
    name = parts[0].strip()
    score_str = parts[1].strip()
    
    # 检查姓名是否为空
    if not name:
        raise ValueError("姓名不能为空")
    
    # 尝试将成绩转换为整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩必须是整数: '{score_str}'")
    
    # 检查成绩是否在 0~100 范围内
    if score < 0 or score > 100:
        raise ValueError(f"成绩必须在 0~100 范围内: {score}")
    
    return name, score

# ========== 测试 ==========

print("=" * 60)
print(" 4: 按行处理文件，捕获异常，收集有效记录")
print("=" * 60)

valid_records = []      # 有效记录：字典组成的列表
invalid_records = []    # 无效记录：(行号, 原始内容, 错误原因)

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

# 从第 2 行开始处理（跳过第 1 行表头）
for line_no, line in enumerate(lines[1:], start=2):
    line = line.strip()
    
    # 跳过空行
    if not line:
        continue
    
    try:
        name, score = parse_student(line)
        # 有效记录保存为字典
        valid_records.append({
            "name": name,
            "score": score
        })
        print(f"✓ 第 {line_no:2d} 行: {name:<8} {score:3d}  [有效]")
        
    except ValueError as e:
        # 捕获异常，记录行号、原始内容和错误原因
        invalid_records.append((line_no, line, str(e)))
        print(f"✗ 第 {line_no:2d} 行: '{line}'  [无效] → {e}")

print("\n" + "=" * 60)
print("处理结果汇总")
print("=" * 60)
print(f"有效记录数: {len(valid_records)}")
print(f"无效记录数: {len(invalid_records)}")

print("\n--- 有效记录（字典列表）---")
for record in valid_records:
    print(record)

print("\n--- 无效记录详情 ---")
for line_no, content, reason in invalid_records:
    print(f"  第 {line_no} 行: '{content}' → {reason}")
    
```

# 实验5:类、继承与对象复制

```python
# ========== 1: 定义 Student 类 ==========

class Student:
    """
    学生类，包含姓名、成绩属性，以及及格判断和字符串表示方法。
    """
    
    def __init__(self, name, score):
       
        self.name = name
        self.score = score
    
    def is_passed(self):
       
        return self.score >= 60
    
    def __str__(self):
     
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name='{self.name}', score={self.score}, 状态={status})"

# 测试 Student 类
print("=" * 60)
print("任务 1: Student 类测试")
print("=" * 60)

s1 = Student("Alice", 85)
s2 = Student("Bob", 58)

print(f"\n学生 1: {s1}")
print(f"是否及格: {s1.is_passed()}")

print(f"\n学生 2: {s2}")
print(f"是否及格: {s2.is_passed()}")
```

```python
# ========== 任务 3: 定义 GraduateStudent 子类 ==========

class GraduateStudent(Student):
    """
    研究生类，继承自 Student，增加研究课题属性。
    """
    
    def __init__(self, name, score, research_topic):
        
        # 使用 super() 调用父类的 __init__ 方法
        super().__init__(name, score)
        self.research_topic = research_topic
    
    def __str__(self):
      
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name='{self.name}', score={self.score}, "
                f"状态={status}, 课题='{self.research_topic}')")

# 测试 GraduateStudent 类
print("\n" + "=" * 60)
print("任务 3: GraduateStudent 子类测试")
print("=" * 60)

g1 = GraduateStudent("Alice", 85, "机器学习")
g2 = GraduateStudent("Bob", 58, "深度学习")

print(f"\n研究生 1: {g1}")
print(f"是否及格: {g1.is_passed()}")

print(f"\n研究生 2: {g2}")
print(f"是否及格: {g2.is_passed()}")

# 验证继承关系
print(f"\n--- 继承关系验证 ---")
print(f"g1 是 Student 的实例: {isinstance(g1, Student)}")
print(f"g1 是 GraduateStudent 的实例: {isinstance(g1, GraduateStudent)}")
```

```python
import copy

# ========== 实验五 任务 4: 对象复制比较 ==========

print("\n" + "=" * 60)
print("任务 4: 直接赋值、copy.copy() 和 copy.deepcopy() 比较")
print("=" * 60)

def create_original():
    """创建原始对象，用于每次测试"""
    return {"name": "Alice", "scores": [85, 90]}

# ---------- 测试 1: 直接赋值 ----------
print("\n--- 测试 1: 直接赋值 (=) ---")
original1 = create_original()
alias1 = original1  # 直接赋值：两个变量指向同一个对象

print(f"修改前:")
print(f"  original: {original1}")
print(f"  alias:    {alias1}")
print(f"  是同一对象: {original1 is alias1}")

# 修改内部列表
alias1["scores"][0] = 100

print(f"\n修改 alias['scores'][0] = 100 后:")
print(f"  original: {original1}")
print(f"  alias:    {alias1}")
#结论: 直接赋值不会创建新对象，修改会影响原对象！

# ---------- 测试 2: 浅拷贝 copy.copy() ----------
print("\n--- 测试 2: 浅拷贝 copy.copy() ---")
original2 = create_original()
shallow2 = copy.copy(original2)  # 浅拷贝：创建新对象，但内部元素共享引用

print(f"修改前:")
print(f"  original: {original2}")
print(f"  shallow:  {shallow2}")
print(f"  是同一对象: {original2 is shallow2}")
print(f"  scores 是同一对象: {original2['scores'] is shallow2['scores']}")

# 修改内部列表
shallow2["scores"][0] = 100

print(f"\n修改 shallow['scores'][0] = 100 后:")
print(f"  original: {original2}")
print(f"  shallow:  {shallow2}")
#浅拷贝创建了新对象，但内部可变对象（如列表）仍共享引用！

# ---------- 测试 3: 深拷贝 copy.deepcopy() ----------
print("\n--- 测试 3: 深拷贝 copy.deepcopy() ---")
original3 = create_original()
deep3 = copy.deepcopy(original3)  # 深拷贝：递归创建所有层级的新对象

print(f"修改前:")
print(f"  original: {original3}")
print(f"  deep:     {deep3}")
print(f"  是同一对象: {original3 is deep3}")
print(f"  scores 是同一对象: {original3['scores'] is deep3['scores']}")

# 修改内部列表
deep3["scores"][0] = 100

print(f"\n修改 deep['scores'][0] = 100 后:")
print(f"  original: {original3}")
print(f"  deep:     {deep3}")
#结论: 深拷贝完全独立，修改不会影响原对象！

# ---------- 汇总比较 ----------
print("\n" + "=" * 60)
print("三种复制方式对比总结")
print("=" * 60)

comparison_data = [
    ["方式", "新对象", "内部元素共享", "修改内部是否影响原对象"],
    ["=", "否", "是", "是"],
    ["copy.copy()", "是", "是", "是"],
    ["copy.deepcopy()", "是", "否", "否"],
]

for row in comparison_data:
    print(f"{row[0]:<20} {row[1]:<10} {row[2]:<15} {row[3]}")
```
