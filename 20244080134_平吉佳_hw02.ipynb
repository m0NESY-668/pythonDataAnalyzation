# 字符串的乐趣
### 回文函数

```python
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

print(is_palindrome("tacocat"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("hello"))
```

### 判断是否按字母顺序排序

```python
def is_abecedarian(s):
    s = s.lower().replace(" ", "")
    for i in range(len(s) - 1):
        if s[i] > s[i + 1]:
            return False
    return True

print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
```

### 去除元音字母函数

```python
def remove_vowels(s):
    result = ""
    for ch in s:
        if ch not in "aeiouAEIOU":
            result += ch
    return result

print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

# 列表的乐趣

```python
def list_reverse(l):
    if not isinstance(l, list):
        raise TypeError("list_reverse() 的参数必须是列表")
    result = []
    for i in range(len(l) - 1, -1, -1):
        result.append(l[i])
    return result

print(list_reverse([1, 2, 3]))
print(list_reverse([]))
print(list_reverse(["a", "b", "c"]))
```

```python
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 3, 2]))
print(is_sorted("abc"))
print(is_sorted([1, 1, 2]))
print(is_sorted([]))
```

# 作为向量的元组
### 实现加法、减法、乘法、除法、点积、模长、归一化等方法

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("vec_scalar_mult() 的第一个参数必须是元组")
    for x in t:
        if not isinstance(x, (int, float)):
            raise TypeError("元组 t 的每个元素都必须是整数或浮点数")
    if not isinstance(s, (int, float)):
        raise TypeError("vec_scalar_mult() 的第二个参数必须是整数或浮点数")
    result = []
    for x in t:
        result.append(x * s)
    return tuple(result)

print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, -2.0), 2))
```

```python
def vec_inner_product(u, v):
    if not isinstance(u, tuple) or not isinstance(v, tuple):
        raise TypeError("vec_inner_product() 的两个参数都必须是元组")
    if len(u) != len(v):
        raise ValueError("两个元组的长度必须相同")
    total = 0.0
    for i in range(len(u)):
        total += u[i] * v[i]
    return total

print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1, 2), (3, 4)))
```

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return False
    if not isinstance(mx[0], tuple):
        return False
    n = len(mx[0])
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != n:
            return False
        for x in row:
            if not isinstance(x, (int, float)):
                return False
    return True

print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), (3, 4, 5))))
print(check_valid_mx(((1, 2), [3, 4])))
print(check_valid_mx(((1, "a"), (3, 4))))
```

```python
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("矩阵无效：必须是元组的元组，各内层元组等长，且元素均为整数或浮点数")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n = len(mx[0])
    if len(vec) != n:
        raise ValueError("向量长度必须等于矩阵的列数")
    result = []
    for row in mx:
        total = 0.0
        for j in range(n):
            total += row[j] * vec[j]
        result.append(total)
    return tuple(result)

print(mx_vec_mult(((1, 2), (3, 4)), (1, 1)))
print(mx_vec_mult(((1, 2, 3), (4, 5, 6)), (1, 2, 3)))
```

# 更多向量乐趣

### 定义vector类

2.维度和元素是实例属性。理由：每个 Vector 实例可以拥有自己独立的维度和元素，不同向量维度、元素各不相同；类属性是属于整个类、所有对象共享的数据，不适合向量这种每个实例独立的数据

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        
        if elements is None:
            elements = [0] * dim
        else:
            elements = tuple(elements)
            if len(elements) != dim:
                raise ValueError(f"维度{dim}和元素数量{len(elements)}不一致")
        
        self.dim = dim
        self.elements = elements

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # ---------- 比较运算符（必须放在class里面！）----------
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements < other.elements

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not self <= other

    def __ge__(self, other):
        return not self < other

    # ---------- dot内积 ----------
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("两个向量维度必须相同才能计算内积")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    # ---------- 乘法运算符 ----------
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_elems = tuple(x * other for x in self.elements)
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量相乘维度必须一致")
            new_elems = tuple(a*b for a,b in zip(self.elements, other.elements))
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("不支持的乘法类型")

    def __rmul__(self, scalar):
        return self * scalar
```

### 测试代码

```python
# 测试初始化
v1 = Vector(2, (2,4))
v2 = Vector(2, (2,5))
v3 = Vector(2, (4,5))
v_zero = Vector(3)  # 不填元素，默认全0向量
print(v1, v2, v_zero)

# 测试比较
print(v1 < v2)
print(v1 == v2)

# 测试内积
print(v1.dot(v2))

# 测试乘法
print(v1 * 2)
print(2 * v1)
print(v1 * v3)
```
